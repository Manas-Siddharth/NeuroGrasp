# 🤖 NeuroGrasp: Deep Learning-Based Robotic Grasp Detection
This notebook demonstrates end-to-end dataset loading, neural network forward pass, model training, metric evaluation, and visual inference for robotic grasp detection.

```python
import os
import sys
import torch
import matplotlib.pyplot as plt

# Add root directory to sys.path
sys.path.append('..')

from src.utils import set_seed, get_device, load_config
from src.dataset import JacquardGraspDataset, create_dataloaders
from src.model import NeuroGraspNet
from src.losses import NeuroGraspLoss
from src.inference import GraspPredictor

set_seed(42)
device = get_device()
print(f"Using device: {device}")
```

## 1. Inspect Sample Dataset

```python
dataset = JacquardGraspDataset(data_dir="../data/sample_dataset", target_size=224)
print(f"Total sample images found: {len(dataset)}")

img_tensor, target_tensor = dataset[0]
print(f"Image tensor shape: {img_tensor.shape}")
print(f"Target tensor values [x, y, w, h, sin_t, cos_t, conf]:\n{target_tensor.numpy()}")
```

## 2. Initialize Neural Network & Smoke Test

```python
model = NeuroGraspNet(backbone_name="resnet18", pretrained=True).to(device)
dummy_batch = torch.randn(2, 3, 224, 224).to(device)
output = model(dummy_batch)
print(f"Batch forward output shape: {output.shape}")
```

## 3. Run Inference on Sample Image

```python
predictor = GraspPredictor(checkpoint_path="../checkpoints/best_model.pth", config_path="../config.yaml")
res = predictor.predict("../data/sample_dataset/1_1_RGB.png")

plt.figure(figsize=(8, 8))
plt.imshow(res['annotated_image'])
plt.title(f"Predicted Grasp (Conf: {res['confidence']:.2f}, Angle: {res['angle_deg']:.1f}°)")
plt.axis('off')
plt.show()
```
